```python
# Qwen3-VL setup for MMMU evaluation (images are PIL in the dataset)
import re
import torch
from tqdm import tqdm


def build_mc_prompt(question: str, options):
    if type(options) is not list:
        options = eval(options)
    letters = [chr(ord('A') + i) for i in range(len(options))]
    opts_str = " ".join([f"({l}) {opt}" for l, opt in zip(letters, options)])
    instr = f"Question: {question}\nOptions: {opts_str}\nAnswer with a single capital letter among {', '.join(letters)} only."
    return instr, letters

def parse_choice(text: str, allowed_letters):
    # Normalize
    if text is None:
        return None
    text_u = str(text).strip().upper()

    # 1) Exact single-letter reply
    if len(text_u) == 1 and text_u in allowed_letters:
        return text_u

    # 2) Regex for standalone single capital letter
    m = re.search(r"\b([A-Z])\b", text_u)
    if m and m.group(1) in allowed_letters:
        return m.group(1)

    # 3) Scan first capital letter token present in allowed set
    for t in re.findall(r"[A-Z]", text_u):
        if t in allowed_letters:
            return t

    # 4) Safe boundary-like match using escaped letter (avoid regex injection)
    for ch in allowed_letters:
        if re.search(rf"(^|[^A-Z]){re.escape(ch)}([^A-Z]|$)", text_u):
            return ch

    return None


def sample_to_images(sample):
    # MMMU provides PIL images; handle common keys
    if "images" in sample and sample["images"]:
        return sample["images"]
    if "image" in sample and sample["image"] is not None:
        img = sample["image"]
        return img if isinstance(img, list) else [img]
    # some tasks may store under other keys; extend if needed
    return []

def gt_to_letter(sample_answer, allowed_letters):
    # ground truth may be a letter like 'A' or index like 0
    if isinstance(sample_answer, str):
        s = sample_answer.strip().upper()
        if len(s) > 0 and s[0] in allowed_letters:
            return s[0]
    if isinstance(sample_answer, int) and 0 <= sample_answer < len(allowed_letters):
        return allowed_letters[sample_answer]
    return None

def eval_subject(ds, subject_name: str):
    correct = 0
    total = 0
    records = []
    for sample in tqdm(ds, desc=f"Eval {subject_name}"):
        question = sample.get("question", "").strip()
        options = sample.get("options", [])
        # images = sample_to_images(sample)
        images= sample['image_1']
        prompt, letters = build_mc_prompt(question, options)
        # messages = build_mm_message(images, prompt)
        # pred_text = qwen_generate(messages)
        ans = bot.inference(prompt, images)[0]
        pred_text = ans.replace("assistant\n", "").strip()
        pred_letter = parse_choice(pred_text, letters) or ""
        gt_letter = gt_to_letter(sample.get("answer"), letters) or ""
        ok = int(pred_letter == gt_letter) if gt_letter else 0
        correct += ok
        total += 1
        records.append({
            "question": question,
            "options": options,
            "pred_text": pred_text,
            "pred": pred_letter,
            "gt": gt_letter,
            "correct": bool(ok)
        })
    acc = correct / total if total > 0 else 0.0
    return acc, records

subject_results = {}
overall_correct = 0
overall_total = 0
for subject, ds in zip(subjects, datasets):
    acc, _ = eval_subject(ds, subject)
    subject_results[subject] = acc
    print(f"{subject}: acc={acc:.4f}")
    overall_correct += int(acc * len(ds))
    overall_total += len(ds)

overall_acc = overall_correct / overall_total if overall_total > 0 else 0.0
print("Overall accuracy:", overall_acc)
```
